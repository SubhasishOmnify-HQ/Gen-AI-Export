# Text Cleaning

```python
# COMPLETE NLP TEXT CLEANING PIPELINE
import re
import unicodedata
import pandas as pd

from bs4 import BeautifulSoup
import emoji

# 1. TEXT CLEANING FUNCTION


def clean_text(text):

    # Check missing values
    if pd.isna(text):
        return ""

    # Convert to string
    text = str(text)

    text = unicodedata.normalize("NFKC", text)

    # 2. Remove HTML Tags

    text = BeautifulSoup(text, "html.parser").get_text(" ")

    # 3. Convert to Lowercase
    text = text.lower()

    # 4. Remove URLs
    text = re.sub(
        r"https?://\S+|www\.\S+",
        " ",
        text
    )


    # 5. Remove Email Addresses
    text = re.sub(
        r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b",
        " ",
        text
    )


    # 6. Remove @Mentions
    # Example: @subhasish → removed

    text = re.sub(
        r"@\w+",
        " ",
        text
    )

    # 7. Handle Hashtags
    # Example: #NLP → NLP
    text = re.sub(
        r"#(\w+)",
        r"\1",
        text
    )

    # 8. Convert Emojis to Text
    # Example: 😍 → smiling_face_with_heart_eyes
    text = emoji.demojize(
        text,
        delimiters=(" ", " ")
    )

    # 9. Remove Emoji Colons
    # Example:
    # :smiling_face_with_heart_eyes:
    # becomes:
    # smiling_face_with_heart_eyes
    text = text.replace(":", " ")

    # 10. Normalize Repeated Characters
    # Example:
    # goooood → good
    # heyyyyy → heyy

    text = re.sub(
        r"(.)\1{2,}",
        r"\1\1",
        text
    )

    # 11. Remove Punctuation
    text = re.sub(
        r"[^\w\s]",
        " ",
        text
    )


    # 12. Remove Underscores
    text = text.replace("_", " ")

    # 13. Handle Numbers
    #
    # We KEEP numbers here because numbers can be
    # important in many NLP tasks.
    #
    # Example:
    # "COVID 19"
    # "iPhone 15"
    # "100%"
    #
    # If your task does not need numbers, use:
    #
    # text = re.sub(r"\d+", " ", text)

    # 14. Remove Extra Whitespace

    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()

    return text



# 2. TEST THE FUNCTION

text = """
    <h1>I LOVE NLP!!! 😍😍</h1>

    NLP is soooo amazing!!!

    Visit https://example.com

    Email: test@example.com

    @Subhasish #ArtificialIntelligence

    I have 100% confidence in NLP.
"""

print("========== ORIGINAL TEXT ==========")
print(text)
```

```python
cleaned = clean_text(text)

print("\n========== CLEANED TEXT ==========")
print(cleaned)
```

# Stopword Handling

### 1. NLTK Implementation

```python
import nltk

nltk.download("stopwords")

from nltk.corpus import stopwords

stop_words = set(
    stopwords.words("english")
)

text = "I am learning natural language processing"

words = text.lower().split()

result = [
    word
    for word in words
    if word not in stop_words
]

print(result)
```

### 2.scikit-learn

```python
from sklearn.feature_extraction.text import TfidfVectorizer

documents = [
    "I am learning NLP",
    "NLP is very interesting",
    "I love machine learning"
]

vectorizer = TfidfVectorizer(
    stop_words="english"
)

X = vectorizer.fit_transform(documents)

print(vectorizer.get_feature_names_out())
```

### 3. Custom Stopwords + TF-IDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer

custom_stopwords = {
    "a", "an", "the",
    "is", "am", "are",
    "was", "were",
    "and", "or",
    "of", "to", "in",
    "on", "for", "with"
}

vectorizer = TfidfVectorizer(
    stop_words=list(custom_stopwords)
)

X = vectorizer.fit_transform(documents)

print(X.shape)
```

# Stemming

### 1.Porter Stemmer

```python
import nltk

from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

words = [
    "play",
    "playing",
    "played",
    "plays",
    "studies",
    "studying",
    "connection",
    "connected"
]

for word in words:
    print(word, "→", stemmer.stem(word))
```

### 2. Snowball Stemmer

```python
from nltk.stem import SnowballStemmer

stemmer = SnowballStemmer("english")

words = [
    "play",
    "playing",
    "played",
    "plays",
    "studies",
    "studying",
    "connection",
    "connected"
]

for word in words:
    print(word, "→", stemmer.stem(word))
```

### 3. Apply Stemming to a Sentence

```python
import nltk

from nltk.tokenize import word_tokenize
from nltk.stem import SnowballStemmer

nltk.download("punkt")

stemmer = SnowballStemmer("english")

text = """
I am playing games.
I played games yesterday.
I love playing different games.
"""

tokens = word_tokenize(text.lower())

stemmed_words = [
    stemmer.stem(word)
    for word in tokens
]

print("Tokens:")
print(tokens)

print("\nStemmed:")
print(stemmed_words)
```

# Lemmatization

### 1.NLTK WordNetLemmatizer

```python
import nltk

nltk.download("wordnet")
nltk.download("omw-1.4")

from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

words = ["running", "studies", "mice", "children"]

for word in words:
    print(word, "→", lemmatizer.lemmatize(word))
```

```python
print(lemmatizer.lemmatize("running", pos="v"))
```

```python
print(lemmatizer.lemmatize("mice", pos="n"))
```

### 2. Stanza

```python
import stanza

stanza.download("en")
nlp = stanza.Pipeline(
    "en",
    processors="tokenize,pos,lemma"
)

text = "The children are running and studying."

doc = nlp(text)

for sentence in doc.sentences:
    for word in sentence.words:
        print(word.text, "→", word.lemma)
```

# N-Grams

```python
from nltk.util import ngrams
from nltk.tokenize import word_tokenize

text = "I love natural language processing"

tokens = word_tokenize(text.lower())

# Unigram
unigrams = list(ngrams(tokens, 1))

# Bigram
bigrams = list(ngrams(tokens, 2))

# Trigram
trigrams = list(ngrams(tokens, 3))

print("Unigrams:")
print(unigrams)

print("\nBigrams:")
print(bigrams)

print("\nTrigrams:")
print(trigrams)
```

# Bag of Words (BoW)

```python
from sklearn.feature_extraction.text import CountVectorizer

documents = [
    "I love NLP",
    "I love machine learning",
    "NLP is interesting"
]

# Create Bag of Words
vectorizer = CountVectorizer()

X = vectorizer.fit_transform(documents)

# Vocabulary
print("Vocabulary:")
print(vectorizer.get_feature_names_out())

# Numerical representation
print("\nBoW Matrix:")
print(X.toarray())
```

# TF-IDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer

documents = [
    "I love machine learning",
    "I love deep learning",
    "Machine learning is powerful"
]

# Create TF-IDF vectorizer
vectorizer = TfidfVectorizer()

# Convert text → numerical matrix
X = vectorizer.fit_transform(documents)

# Vocabulary
print("Vocabulary:")
print(vectorizer.get_feature_names_out())

# TF-IDF matrix
print("\nTF-IDF Matrix:")
print(X.toarray())
```

## Get TF-IDF Values Clearly

```python
import pandas as pd

feature_names = vectorizer.get_feature_names_out()

tfidf_df = pd.DataFrame(
    X.toarray(),
    columns=feature_names
)

print(tfidf_df)
```
